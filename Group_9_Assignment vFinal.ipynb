```python
#Best win streak of consecutive races of Alonso
races=list(fAlonso["race_date"])
alonso_races=sorted(races)
streak=0
win_streak=0
dates=[]
win_dates=[]
for date in alonso_races:
    final_position=fAlonso[fAlonso["race_date"]==date]["final_position"].iloc[0]
    if final_position==1:
        streak+=1
        dates.append(date)
    else:
        if win_streak<streak:
            win_streak=streak
            streak=0
            win_dates=dates
            dates=[]
        else:
            streak=0
            dates=[]

win_streak,win_dates
print(f"Win streak: {win_streak}")
print(f"Dates: {win_dates}")
```

```python
#Best podium streak of consecutive races of Alonso
races=list(fAlonso["race_date"])
alonso_races=sorted(races)
streak=0
podium_streak=0
dates=[]
podium_dates=[]
for date in alonso_races:
    final_position=fAlonso[fAlonso["race_date"]==date]["final_position"].iloc[0]
    if final_position==1 or final_position==2 or final_position==3:
        streak+=1
        dates.append(date)
    else:
        if podium_streak<streak:
            podium_streak=streak
            streak=0
            podium_dates=dates
            dates=[]
        else:
            streak=0
            dates=[]

win_streak,win_dates
print(f"Podium streak: {podium_streak}")
print(f"Dates: {podium_dates}")
```

```python
#Separate all the drivers by their different seasons--> It takes a while to compile
all_years=set(f1data["year"])
all_drivers=set(f1data["driver"])
all_drivers_sesons={}
for driver in all_drivers:
    all_drivers_sesons[driver]={}
    for year in all_years:
        all_drivers_sesons[driver][year]=f1data[(f1data["driver"]==driver) & (f1data["year"]==year)]
```

```python
#Create a dictionary for each driver with the points obtained in each season
all_drivers_points={}
for driver in all_drivers:
    all_drivers_points[driver]={}
    for year in all_years:
        all_drivers_points[driver][year]=all_drivers_sesons[driver][year]["points"].sum()
```

```python
#Get the ranking of the drivers in alonso years
ranking_alonso_years={}
ranking={}
for year in alonso_years:
    for driver in all_drivers:
        ranking[driver]=all_drivers_points[driver][year]
    ranking_alonso_years[year]=sorted(ranking.items(),key=lambda x:x[1],reverse=True)
```

```python
#Find in which years Alonso won the championship
years_Alonso_won=[]
for year,rank in ranking_alonso_years.items():
    if rank[0][0]=="Fernando Alonso":
        years_Alonso_won.append(year)
years_Alonso_won
```
